```python
my_age=int(input('Enter my age: '))
your_age=int(input('Enter your age: '))

if my_age>=0 and your_age>=0:
    if your_age-my_age==1:
        print('Your are a year older than me')
    elif your_age-my_age>1: 
        dif=str(your_age-my_age)
        print('You are ' + dif + ' years older than me')
    elif your_age==my_age: 
        print('We are the same age')
        
else:
    print('You are older than me')
    
```

```python
numbera=int(input('Enter number one: '))
numberb=int(input('Enter number two: '))

if numbera==numberb:
    numbera=str(numbera)
    numberb=str(numberb)
    print(numbera + ' is equal to ' + numberb)
elif numbera>numberb:
    numbera=str(numbera)
    numberb=str(numberb)
    print(numbera + ' is greater than ' + numberb)
elif numbera<numberb:
    numbera=str(numbera)
    numberb=str(numberb)
    print(numbera + ' is less than ' + numberb)
```

```python
grade=int(input('What grade did you get? '))

if grade>89:
    print('You got a A')
elif grade>79 and grade<90:
    print('You got a B')    
elif grade>69 and grade<80:
    print('You got a C')   
elif grade>49 and grade<60:
    print('You got a D')   
else:
    print('You got a F')   
```

```python
month=(input('What month is it? ')).lower()

if month in ('december', 'january', 'february'):
    print('Winter')
elif month in ('march', 'april', 'may'):
    print('Spring')
elif month in ('june', 'july', 'august'):
    print('Summer')
elif month in ('september', 'october', 'november'):
    print('Autumn')
else:
    print('Not in month list')
```
